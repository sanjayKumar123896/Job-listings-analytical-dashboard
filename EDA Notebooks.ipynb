```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully.")
```

```python
from pathlib import Path
import pandas as pd

# Find the project root automatically
current_path = Path.cwd()

project_root = None

for path in [current_path] + list(current_path.parents):
    if (path / "data" / "processed" / "job_listings_processed.csv").exists():
        project_root = path
        break

if project_root is None:
    raise FileNotFoundError(
        "Could not find job_listings_processed.csv"
    )

# Dataset path
data_path = (
    project_root
    / "data"
    / "processed"
    / "job_listings_processed.csv"
)

# Load dataset
df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
print("Dataset location:", data_path)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
job_category_counts = df["job_category"].value_counts()

plt.figure(figsize=(10, 6))

job_category_counts.plot(
    kind="bar"
)

plt.title("Job Distribution by Category")
plt.xlabel("Job Category")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Most job listings fall under the Other category, while Data & Analytics and Sales & Business have fewer job listings.

```python
experience_counts = df["experience_level"].value_counts()

plt.figure(figsize=(8, 5))

experience_counts.plot(
    kind="bar"
)

plt.title("Job Distribution by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Most job listings are for Mid Level / Unspecified positions, while Entry Level positions have the fewest listings.

```python
location_counts = (
    df["location"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10, 6))

location_counts.plot(
    kind="bar"
)

plt.title("Top 10 Job Locations")
plt.xlabel("Location")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The job listings are evenly distributed across the top 10 locations, with each location having one job listing.

```python
company_counts = (
    df["company_name"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10, 6))

company_counts.plot(
    kind="bar"
)

plt.title("Top 10 Hiring Companies")
plt.xlabel("Company")
plt.ylabel("Number of Job Listings")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Garcia PLC has the highest number of job listings, while the other top companies have fewer listings in the dataset.

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["description_word_count"],
    bins=15
)

plt.title("Distribution of Job Description Word Count")
plt.xlabel("Number of Words")
plt.ylabel("Number of Jobs")

plt.tight_layout()
plt.show()
```

Most job descriptions contain around 40–55 words, while fewer descriptions have very high word counts.

```python
skill_columns = [
    "python_required",
    "sql_required",
    "machine_learning_required",
    "ai_required",
    "aws_required",
    "java_required",
    "javascript_required",
    "docker_required",
    "cloud_required"
]

skill_counts = (
    df[skill_columns]
    .sum()
    .sort_values(ascending=False)
)

skill_counts.index = (
    skill_counts.index
    .str.replace("_required", "", regex=False)
    .str.replace("_", " ", regex=False)
    .str.title()
)

print(skill_counts)
```

```python
plt.figure(figsize=(10, 6))

skill_counts.plot(
    kind="bar"
)

plt.title("Demand for Technical Skills")
plt.xlabel("Skill")
plt.ylabel("Number of Job Listings")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Python is the most demanded technical skill in the dataset, followed by Java, while the other skills have lower demand.

```python


company_counts = (
    df["company_name"]
    .value_counts()
    .head(10)
    .sort_values(ascending=True)
)

plt.figure(figsize=(10, 6))

company_counts.plot(
    kind="barh"
)

plt.xlabel("Number of Jobs")
plt.ylabel("Company Name")
plt.title("Top 10 Companies by Job Listings")

plt.tight_layout()
plt.show()
```

Garcia PLC has the highest number of job listings, while the other top companies have relatively similar hiring demand.

```python
category_experience = pd.crosstab(
    df["job_category"],
    df["experience_level"]
)

print(category_experience)
```

```python
category_experience.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Job Category vs Experience Level")
plt.xlabel("Job Category")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Mid-level or unspecified jobs dominate most categories, while senior-level jobs are comparatively fewer.

```python
experience_description = (
    df.groupby("experience_level")[
        "description_word_count"
    ]
    .mean()
    .sort_values(ascending=False)
)

print(experience_description)
```

```python
plt.figure(figsize=(8, 5))

experience_description.plot(
    kind="bar"
)

plt.title(
    "Average Job Description Length by Experience Level"
)

plt.xlabel("Experience Level")
plt.ylabel("Average Word Count")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Entry-level jobs have the longest average descriptions, while mid-level or unspecified jobs have the comparatively shortest average descriptions.

```python
numeric_df = df.select_dtypes(
    include=np.number
)

correlation = numeric_df.corr()

correlation
```

```python
plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title(
    "Correlation Heatmap of Numerical Features"
)

plt.tight_layout()
plt.show()
```

The heatmap shows a strong positive correlation between description word count and character count, while Python and Java requirements also show a positive relationship with description length.
